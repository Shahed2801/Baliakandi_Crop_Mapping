```python
import fitz # PyMuPDF
from google.colab import files

input_pdf_path = '/content/Google_Earth_Engine_Applications.pdf'
output_pdf_path = 'Google_Earth_Engine_Applications_Split.pdf'

# Open the input PDF
doc = fitz.open(input_pdf_path)

# Create a new PDF document
new_doc = fitz.open()

# Page numbers are 0-indexed in fitz. User requested page 16 to 406.
# So, in 0-indexed, it's from page 15 up to (but not including) page 406.
# The range will be [start_page_index, end_page_index + 1]
start_page_index = 15  # Page 16 in 1-indexed
end_page_index = 405    # Page 406 in 1-indexed

# Check if the requested page range is valid
if start_page_index < 0 or end_page_index >= len(doc):
    print(f"Error: Page range is out of bounds. The document has {len(doc)} pages.")
else:
    # Insert pages from the original document into the new document
    new_doc.insert_pdf(doc, from_page=start_page_index, to_page=end_page_index)

    # Save the new PDF
    new_doc.save(output_pdf_path)
    new_doc.close()
    doc.close()

    print(f"Successfully created '{output_pdf_path}' with pages {start_page_index + 1} to {end_page_index + 1}.")

    # Provide a download link for the new PDF
    files.download(output_pdf_path)
```
